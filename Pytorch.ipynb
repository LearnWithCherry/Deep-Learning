```python
import torch
import time

device = "cuda"

# Large matrices
size = 5000

a = torch.randn(size, size, device=device)
b = torch.randn(size, size, device=device)

# Warm-up
for _ in range(5):
    c = torch.mm(a, b)

torch.cuda.synchronize()

# Benchmark
start = time.perf_counter()

for _ in range(10):
    c = torch.mm(a, b)

torch.cuda.synchronize()
elapsed = time.perf_counter() - start

print(f"GPU: {torch.cuda.get_device_name(0)}")
print(f"Time for 10 matrix multiplications: {elapsed:.2f} seconds")
print(f"Average: {elapsed/10:.3f} seconds per operation")
```

```python
import torch
import time

size = 5000
iterations = 20

a = torch.randn(size, size, device="cuda")
b = torch.randn(size, size, device="cuda")

# Warm-up
for _ in range(5):
    torch.mm(a, b)

torch.cuda.synchronize()

start = time.perf_counter()

for _ in range(iterations):
    torch.mm(a, b)

torch.cuda.synchronize()

elapsed = time.perf_counter() - start

flops = 2 * size**3 * iterations
tflops = flops / elapsed / 1e12

print(f"GPU: {torch.cuda.get_device_name(0)}")
print(f"Total time: {elapsed:.3f} seconds")
print(f"Average: {elapsed/iterations*1000:.2f} ms")
print(f"Approx. FP32 throughput: {tflops:.2f} TFLOPS")
```

```python
import torch
import time

# Make sure CUDA is available
assert torch.cuda.is_available(), "CUDA GPU not available!"

device = torch.device("cuda")
print("GPU:", torch.cuda.get_device_name(0))

# Large matrices
size = 6000

a = torch.randn(size, size, device=device)
b = torch.randn(size, size, device=device)

# Warm up GPU
for _ in range(5):
    torch.mm(a, b)

torch.cuda.synchronize()

print("GPU stress test started...")

start = time.perf_counter()
operations = 0

while time.perf_counter() - start < 10:
    c = torch.mm(a, b)
    operations += 1

# Make sure all GPU work has finished
torch.cuda.synchronize()

elapsed = time.perf_counter() - start

print("GPU stress test finished!")
print(f"GPU: {torch.cuda.get_device_name(0)}")
print(f"Duration: {elapsed:.2f} seconds")
print(f"Matrix multiplications: {operations}")
```
